# Demo 1 — Le particelle oscillano, la perturbazione si sposta

Modificare i parametri all’inizio della cella di codice ed eseguirla con **Shift + Invio**. Per avviare tutto senza selezionare la cella: **Run → Run All Cells**.

Questa demo è indipendente dalle altre e viene calcolata nel browser. Il primo avvio può richiedere qualche istante.

```python
# Parametri da modificare
frequenza = 2.0         # Hz: modello rallentato per osservare il moto
istante = 0.0          # secondi
passo_temporale = 0.1   # secondi fra le due istantanee

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

FS = 44100
plt.rcParams.update({"figure.figsize": (9, 3), "axes.grid": True, "grid.alpha": 0.25})

def tempo(durata):
    assert 0 < durata <= 10, "Usare una durata fra 0 e 10 secondi."
    return np.arange(round(FS * durata)) / FS

def ascolto(y, etichetta):
    y = np.asarray(y, dtype=float).copy()
    assert np.max(np.abs(y)) <= 1, "Picco oltre fondo scala: ridurre le ampiezze."
    n = min(round(0.01 * FS), len(y) // 2)
    if n:
        rampa = np.linspace(0, 1, n)
        y[:n] *= rampa
        y[-n:] *= rampa[::-1]
    print(etichetta)
    display(Audio(y, rate=FS, normalize=False))

print("Python pronto: esecuzione della demo.")


assert 0.2 <= frequenza <= 5 and 0 <= istante <= 10 and 0 < passo_temporale <= 1
c = 2.0                # velocità del modello, m/s (non dell’aria reale)
x0 = np.linspace(0, 4, 65)
x = np.linspace(0, 4, 1000)
k = 2 * np.pi * frequenza / c
fig, axes = plt.subplots(2, 1, figsize=(9, 4), sharex=True)
for ax, t in zip(axes, [istante, istante + passo_temporale]):
    # Spostamento ingrandito; la pressione è legata alla compressione del mezzo.
    spostamento = 0.025 * np.sin(k * x0 - 2 * np.pi * frequenza * t)
    pressione = -np.cos(k * x - 2 * np.pi * frequenza * t)
    ax.vlines(x0, -0.1, 0.1, color="gray", alpha=0.3)
    ax.scatter(x0 + spostamento, np.zeros_like(x0), s=20, label="Particelle")
    ax.plot(x, 0.35 * pressione + 0.65, color="#d9362b", label="Pressione (scala grafica)")
    ax.set_title(f"Istante {t:.2f} s"); ax.set_ylim(-0.2, 1.1)
    ax.set_ylabel("Schema"); ax.legend(loc="upper right", fontsize=8)
axes[-1].set_xlabel("Posizione [m, modello rallentato]")
plt.tight_layout(); plt.show(); plt.close()
print("Le particelle oscillano vicino ai riferimenti grigi; la perturbazione avanza.")
```
