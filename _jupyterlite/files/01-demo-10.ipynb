Questa demo è indipendente dalle altre e viene calcolata nel browser. Il primo avvio può richiedere qualche istante.

```python
# Parametri da modificare
potenza = 1.0             # W acustici, valore didattico
distanza_riferimento = 1.0  # m

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

FS = 44100
plt.rcParams.update({"figure.figsize": (9, 3), "axes.grid": True, "grid.alpha": 0.25})

def tempo(durata):
    assert 0 < durata <= 10, "Usare una durata fra 0 e 10 secondi."
    return np.arange(round(FS * durata)) / FS

def ascolto(y, etichetta):
    y = np.asarray(y, dtype=float).copy()
    assert np.max(np.abs(y)) <= 1, "Picco oltre fondo scala: ridurre le ampiezze."
    n = min(round(0.01 * FS), len(y) // 2)
    if n:
        rampa = np.linspace(0, 1, n)
        y[:n] *= rampa
        y[-n:] *= rampa[::-1]
    print(etichetta)
    display(Audio(y, rate=FS, normalize=False))

print("Python pronto: esecuzione della demo.")


assert 0 < potenza <= 100 and 0.1 <= distanza_riferimento <= 100
r = np.logspace(-1, 2, 500)
I = potenza / (4 * np.pi * r**2)
I_rif = potenza / (4 * np.pi * distanza_riferimento**2)
fig, axes = plt.subplots(1, 2, figsize=(9, 3))
axes[0].loglog(r, I)
axes[0].set(xlabel="Distanza [m]", ylabel="Intensità [W/m²]", title="Potenza distribuita sulla sfera")
axes[1].semilogx(r, 10 * np.log10(I / I_rif), color="#d9362b")
axes[1].set(xlabel="Distanza [m]", ylabel="Variazione [dB]", title=f"Riferimento: {distanza_riferimento:g} m")
plt.tight_layout(); plt.show(); plt.close()
for fattore in (1, 2, 4, 8):
    d = distanza_riferimento * fattore
    print(f"r = {d:g} m: intensità {potenza/(4*np.pi*d*d):.6f} W/m²; variazione {-20*np.log10(fattore):.2f} dB")
print("Modello: sorgente puntiforme omnidirezionale in campo libero.")
```
