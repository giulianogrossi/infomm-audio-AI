Questa demo è indipendente dalle altre e viene calcolata nel browser. Il primo avvio può richiedere qualche istante.

```python
# Parametri da modificare
rapporto_pressione = 2.0       # provare 0.5, 2, 10
pressione_riferimento = 20e-6  # Pa

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

FS = 44100
plt.rcParams.update({"figure.figsize": (9, 3), "axes.grid": True, "grid.alpha": 0.25})

def tempo(durata):
    assert 0 < durata <= 10, "Usare una durata fra 0 e 10 secondi."
    return np.arange(round(FS * durata)) / FS

def ascolto(y, etichetta):
    y = np.asarray(y, dtype=float).copy()
    assert np.max(np.abs(y)) <= 1, "Picco oltre fondo scala: ridurre le ampiezze."
    n = min(round(0.01 * FS), len(y) // 2)
    if n:
        rampa = np.linspace(0, 1, n)
        y[:n] *= rampa
        y[-n:] *= rampa[::-1]
    print(etichetta)
    display(Audio(y, rate=FS, normalize=False))

print("Python pronto: esecuzione della demo.")


assert rapporto_pressione > 0 and pressione_riferimento > 0
rapporto_intensita = rapporto_pressione**2
variazione = 20 * np.log10(rapporto_pressione)
print(f"Rapporto di pressione: {rapporto_pressione:g}; rapporto di intensità: {rapporto_intensita:g}")
print(f"20 log10(r_p) = 10 log10(r_I) = {variazione:.2f} dB (stesso mezzo, onda progressiva)")
rapporti = np.logspace(-2, 3, 500)
plt.semilogx(rapporti, 20 * np.log10(rapporti))
plt.scatter([rapporto_pressione], [variazione], color="#d9362b", zorder=3)
plt.xlabel("Rapporto di pressione"); plt.ylabel("Differenza di livello [dB]")
plt.title("Scala logaritmica dei decibel")
plt.tight_layout(); plt.show(); plt.close()
pressione = pressione_riferimento * rapporto_pressione
print(f"Con riferimento {pressione_riferimento:g} Pa, la pressione è {pressione:g} Pa.")
```
