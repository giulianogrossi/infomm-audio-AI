# Demo 12 — Una sorgente, due sorgenti non correlate

Modificare i parametri all’inizio della cella di codice ed eseguirla con **Shift + Invio**. Per avviare tutto senza selezionare la cella: **Run → Run All Cells**.

Questa demo è indipendente dalle altre e viene calcolata nel browser. Il primo avvio può richiedere qualche istante.

```python
# Parametri da modificare
numero_sorgenti = 2         # intero fra 1 e 16
seed = 12

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

FS = 44100
plt.rcParams.update({"figure.figsize": (9, 3), "axes.grid": True, "grid.alpha": 0.25})

def tempo(durata):
    assert 0 < durata <= 10, "Usare una durata fra 0 e 10 secondi."
    return np.arange(round(FS * durata)) / FS

def ascolto(y, etichetta):
    y = np.asarray(y, dtype=float).copy()
    assert np.max(np.abs(y)) <= 1, "Picco oltre fondo scala: ridurre le ampiezze."
    n = min(round(0.01 * FS), len(y) // 2)
    if n:
        rampa = np.linspace(0, 1, n)
        y[:n] *= rampa
        y[-n:] *= rampa[::-1]
    print(etichetta)
    display(Audio(y, rate=FS, normalize=False))

print("Python pronto: esecuzione della demo.")


assert isinstance(numero_sorgenti, int) and 1 <= numero_sorgenti <= 16
rng = np.random.default_rng(seed)
# Segnali indipendenti di uguale potenza attesa: la stima finita è approssimata.
sorgenti = rng.standard_normal((numero_sorgenti, FS)) * 0.025
singola = sorgenti[0]
somma = sorgenti.sum(axis=0)
previsto = 10 * np.log10(numero_sorgenti)
misurato = 10 * np.log10(np.mean(somma**2) / np.mean(singola**2))
print(f"Aumento previsto: {previsto:.2f} dB; stimato sui campioni: {misurato:.2f} dB")
plt.bar(["Una sorgente", f"{numero_sorgenti} sorgenti"], [np.mean(singola**2), np.mean(somma**2)])
plt.ylabel("Potenza media digitale"); plt.title("Le sorgenti non correlate sommano le potenze medie")
plt.tight_layout(); plt.show(); plt.close()
ascolto(singola, "Una sorgente")
ascolto(somma, "Somma, con lo stesso guadagno di ascolto")
```
