```python
# Parametri da modificare
frequenza = 200.0       # Hz
temperatura = 20.0      # °C

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

FS = 44100
plt.rcParams.update({"figure.figsize": (9, 3), "axes.grid": True, "grid.alpha": 0.25})

def tempo(durata):
    assert 0 < durata <= 10, "Usare una durata fra 0 e 10 secondi."
    return np.arange(round(FS * durata)) / FS

def ascolto(y, etichetta):
    y = np.asarray(y, dtype=float).copy()
    assert np.max(np.abs(y)) <= 1, "Picco oltre fondo scala: ridurre le ampiezze."
    n = min(round(0.01 * FS), len(y) // 2)
    if n:
        rampa = np.linspace(0, 1, n)
        y[:n] *= rampa
        y[-n:] *= rampa[::-1]
    print(etichetta)
    display(Audio(y, rate=FS, normalize=False))

print("Python pronto: esecuzione della demo.")


assert 20 <= frequenza <= 5000 and -20 <= temperatura <= 50
c = 331.3 + 0.606 * temperatura  # approssimazione per l’aria
periodo = 1 / frequenza
lunghezza_onda = c / frequenza
t = np.linspace(0, 0.02, 2000)
x = np.linspace(0, 4, 2000)
fig, axes = plt.subplots(1, 2, figsize=(9, 3))
axes[0].plot(t * 1000, np.sin(2 * np.pi * frequenza * t))
axes[0].set(xlabel="Tempo [ms]", ylabel="Ampiezza", title="Oscillazione nel tempo")
axes[1].plot(x, np.sin(2 * np.pi * x / lunghezza_onda))
axes[1].set(xlabel="Distanza [m]", title="Istantanea nello spazio")
plt.tight_layout(); plt.show(); plt.close()
print(f"Velocità: {c:.1f} m/s; periodo: {periodo*1000:.2f} ms; lunghezza d’onda: {lunghezza_onda:.3f} m")
```
