# Demo 4 — Separare gli effetti dei tre parametri

Modificare i parametri all’inizio della cella di codice ed eseguirla con **Shift + Invio**. Per avviare tutto senza selezionare la cella: **Run → Run All Cells**.

Questa demo è indipendente dalle altre e viene calcolata nel browser. Il primo avvio può richiedere qualche istante.

```python
# Parametri da modificare: un cambiamento per ogni confronto
A = 0.5                  # ampiezza di riferimento
frequenza = 5.0          # Hz, esempio visivo rallentato
fattore_ampiezza = 2.0
fattore_frequenza = 2.0
sfasamento_gradi = 90.0

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

FS = 44100
plt.rcParams.update({"figure.figsize": (9, 3), "axes.grid": True, "grid.alpha": 0.25})

def tempo(durata):
    assert 0 < durata <= 10, "Usare una durata fra 0 e 10 secondi."
    return np.arange(round(FS * durata)) / FS

def ascolto(y, etichetta):
    y = np.asarray(y, dtype=float).copy()
    assert np.max(np.abs(y)) <= 1, "Picco oltre fondo scala: ridurre le ampiezze."
    n = min(round(0.01 * FS), len(y) // 2)
    if n:
        rampa = np.linspace(0, 1, n)
        y[:n] *= rampa
        y[-n:] *= rampa[::-1]
    print(etichetta)
    display(Audio(y, rate=FS, normalize=False))

print("Python pronto: esecuzione della demo.")


assert 0 < A <= 1 and 0.5 <= frequenza <= 20
assert 0.1 <= fattore_ampiezza <= 3 and 0.2 <= fattore_frequenza <= 4
t = np.linspace(0, 1, 2000)
riferimento = A * np.sin(2 * np.pi * frequenza * t)
varianti = [A * fattore_ampiezza * np.sin(2 * np.pi * frequenza * t),
            A * np.sin(2 * np.pi * frequenza * fattore_frequenza * t),
            A * np.sin(2 * np.pi * frequenza * t + np.deg2rad(sfasamento_gradi))]
fig, axes = plt.subplots(3, 1, figsize=(9, 5), sharex=True)
for ax, y, titolo in zip(axes, varianti, ["Solo ampiezza", "Solo frequenza", "Solo fase"]):
    ax.plot(t, riferimento, label="Riferimento", alpha=0.5)
    ax.plot(t, y, label="Variante", linewidth=1.5)
    ax.set_ylim(-3.2, 3.2); ax.set_ylabel("Ampiezza"); ax.set_title(titolo)
    ax.legend(loc="upper right")
axes[-1].set_xlabel("Tempo [s]")
plt.tight_layout(); plt.show(); plt.close()
print("Le frequenze sono rallentate per leggere i tre confronti: questa demo è visiva.")
```
