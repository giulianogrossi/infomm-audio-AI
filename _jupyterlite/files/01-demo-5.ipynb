Questa demo è indipendente dalle altre e viene calcolata nel browser. Il primo avvio può richiedere qualche istante.

```python
# Parametri da modificare
frequenza_riferimento = 440.0
intervalli_semitoni = (-12, 0, 12)  # un’ottava sotto, riferimento, un’ottava sopra
ampiezza = 0.18

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

FS = 44100
plt.rcParams.update({"figure.figsize": (9, 3), "axes.grid": True, "grid.alpha": 0.25})

def tempo(durata):
    assert 0 < durata <= 10, "Usare una durata fra 0 e 10 secondi."
    return np.arange(round(FS * durata)) / FS

def ascolto(y, etichetta):
    y = np.asarray(y, dtype=float).copy()
    assert np.max(np.abs(y)) <= 1, "Picco oltre fondo scala: ridurre le ampiezze."
    n = min(round(0.01 * FS), len(y) // 2)
    if n:
        rampa = np.linspace(0, 1, n)
        y[:n] *= rampa
        y[-n:] *= rampa[::-1]
    print(etichetta)
    display(Audio(y, rate=FS, normalize=False))

print("Python pronto: esecuzione della demo.")


frequenze = [frequenza_riferimento * 2**(s / 12) for s in intervalli_semitoni]
assert len(frequenze) == 3 and all(20 <= f <= 5000 for f in frequenze)
assert 0 <= ampiezza <= 0.5
plt.bar([f"{s:+g} semitoni" for s in intervalli_semitoni], frequenze)
plt.ylabel("Frequenza [Hz]"); plt.title("Gli intervalli musicali sono rapporti di frequenza")
plt.tight_layout(); plt.show(); plt.close()
for f, s in zip(frequenze, intervalli_semitoni):
    print(f"{f:g} Hz: rapporto {f/frequenza_riferimento:g}, intervallo {s:g} semitoni")
    ascolto(ampiezza * np.sin(2 * np.pi * f * tempo(1)), f"{f:g} Hz")
```
